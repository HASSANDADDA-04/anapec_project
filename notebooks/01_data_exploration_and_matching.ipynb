# 🇲🇦 ANAPEC — Exploration des Données & Moteur de Matching Sémantique

Ce notebook présente la démarche méthodologique de la plateforme **ANAPEC** :
1. Chargement et inspection des données normalisées (**ESCO v1.2.1**).
2. Analyse de la répartition par secteur d'activité (12 domaines clés).
3. Démonstration du matching sémantique par embeddings vectoriels.
4. Optimisation et validation du seuil F1-score.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import json

# Chemins normalisés Cookiecutter Data Science
ROOT_DIR = Path("..").resolve()
DATA_RAW = ROOT_DIR / "data" / "raw" / "esco"
DATA_PROCESSED = ROOT_DIR / "data" / "processed"
MODELS_DIR = ROOT_DIR / "models" / "matching"
REPORTS_DIR = ROOT_DIR / "reports" / "metrics"

print(f"Racine du projet : {ROOT_DIR}")
```

## 1. Chargement des Données Préparées

```python
occupations = pd.read_csv(DATA_PROCESSED / "occupations.csv")
skills = pd.read_csv(DATA_PROCESSED / "skills.csv")
relations = pd.read_csv(DATA_PROCESSED / "relations.csv")

print(f"Nombre de métiers : {len(occupations):,}")
print(f"Nombre de compétences : {len(skills):,}")
print(f"Nombre de relations métier-compétence : {len(relations):,}")
```

## 2. Répartition des Métiers par Domaine d'Activité

```python
domain_counts = occupations["domain"].value_counts()
domain_counts.to_frame("Nombre de métiers")
```

## 3. Consultation des Métriques du Modèle Entraîné

```python
with open(MODELS_DIR / "trained.json", "r", encoding="utf-8") as f:
    model_metrics = json.load(f)

print("Paramètres et performances du modèle optimisé :")
print(json.dumps(model_metrics, indent=2, ensure_ascii=False))
```
